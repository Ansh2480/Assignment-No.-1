```python
print("LIST OPERATION")
student_list = ["Ansh","Aryan"]
print("original list:",student_list)

student_list.append("Jack")
print("After Add (Append 'jack'):",student_list)

student_list[1]="piyush"
print("After Update ('Index 1 to piyush'):",student_list)

student_list.remove("Ansh")
print("After Delete(Removed 'Ansh'):",student_list)
print("-"*30)
```

```python
print("TUPLE OPERATION")
student_tuple=("Ansh","Aryan")
print("original tuple:",student_tuple)

student_tuple=student_tuple + ("jack",)
print("After Add (concatenated'jack'):",student_tuple)

temp_list=list(student_tuple)
temp_list[1]="piyush"
student_tuple=tuple(temp_list)
print("After Update ('jack' to 'piyush'):",student_tuple)


temp_list=list(student_tuple)
temp_list.remove("Ansh")
student_tuple=tuple(temp_list)
print("AfterDelete(Removed 'Ansh'):",student_tuple)
print("-"*30)
```

```python
print("DICTIONARY OPERATION")
student_dict = {
    101:"Ansh",
    102:"Aryan",
}
print("Original Dictionary:",student_dict)

student_dict[103]="piyush"
print("After Add(ID 103:'piyush'):",student_dict)

student_dict[101]="jack"
print("After Update (ID 101 to 'jack'):",student_dict)

del student_dict[102]
print("After Delete(Removed ID 103):",student_dict)
print("-"*30)
```
